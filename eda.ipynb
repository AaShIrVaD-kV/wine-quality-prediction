```python
#visulisation
sns.countplot(x='quality', data=df)
plt.figsize=(10,5)
plt.show()
```

```python
#correlation subplot of target column vs all other columns
plt.figure(figsize=(10,5))
sns.heatmap(df.corr()[['quality']].sort_values(by='quality', ascending=False), annot=True, cmap='viridis')
plt.show()
```

```python
#heatmap of all columns
plt.figure(figsize=(10,5))
sns.heatmap(df.corr(), annot=True, cmap='viridis')
plt.show()
```

```python
#correlation calculation and visulisation of all columns with target column
correlation = df.corr()['quality'].sort_values(ascending=False)
plt.figure(figsize=(10,5))
sns.barplot(x=correlation.index, y=correlation.values)
plt.xticks(rotation=90)
plt.show()
```
